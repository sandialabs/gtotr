# PToTR Paper Case Study 01: Tensor Autoregression with the Integrated Crisis Early Warning System (ICEWS) Dataset 

## Data Processing Notebook

This notebook documents the construction of the `icews_tensor_data.tns` dataset used in the tensor autoregression analysis. The notebook provides a reproducible implementation of the data-processing procedure described in Peter Hoff, “Multilinear tensor regression for longitudinal relational data,” *The Annals of Applied Statistics*, vol. 9, no. 3, pp. 1169–1193, November 2015.

The resulting dataset, `icews_tensor_data.tns`, is a four-way tensor of dimension

$$
25 \times 25 \times 4 \times 548,
$$

with dimensions corresponding to:

- 25 sender countries;
- 25 receiving countries;
- 4 CAMEO event categories (`V+`, `V-`, `M+`, and `M-`); and
- 548 consecutive weeks, covering the period from Thursday, January 1,
  2004 through Wednesday, July 2, 2014.

## Tensor creation:

````python
from pathlib import Path
from ptotr_paper_functions import build_icews_tensor

DATA_DIR = Path("data")
Ytensor, metadata = build_icews_tensor(DATA_DIR)

assert Ytensor.shape == (25, 25, 4, 548)

print(f"Tensor dimensions: {Ytensor.shape}")
print(f"Number of weeks: {Ytensor.shape[3]}")
print(f"Time range: {metadata['week_starts'][0].date()} to {metadata['week_starts'][-1].date()}")

print("\nFirst 10 weeks:")
for week, total in zip(
    metadata["week_starts"][:10],
    metadata["weekly_totals"][:10],
):
    print(f"  {week.date()}: {total:,} events")
````

# Saving the tensor

````python
import pyttb as ttb
import numpy as np

nz = np.nonzero(Ytensor)
subs = np.column_stack(nz)
vals = Ytensor[nz]
Ytensor_spten = ttb.sptensor(subs, vals, Ytensor.shape)
ttb.export_data(Ytensor_spten,"data/icews_tensor_data.tns")
````
